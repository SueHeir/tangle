# Periodic domain-size sweep for DIRT

Layered stacks of one wavy fiber type (length 10, three out-of-plane waves so layers interlock) in cells that are periodic in x and y, with the in-plane side swept from 10 down to 1. Areal density, layer count, and final volume fraction are fixed, so every specimen has the same nominal thickness and the fiber count scales with the cell area. Each configuration is relaxed, compacted, polished for DEM contact, and exported as spherocylinder and bond data for a DIRT through-thickness tension test.

```python
import periodic_domain_sweep as example

for side in example.DOMAIN_SIDES:
    config = example.SweepConfig(side=side)
    print(
        f"side {side:>2}: {config.fiber_count:>3} fibers x {config.segments_per_fiber} segments "
        f"= {config.segment_count:>4} segments, target thickness {config.target_thickness:.2f}"
    )
```

```python
recipe, settings, config = example.build(10.0)
print(f"{len(recipe.operations())} recipe operations")
print(*recipe.operations()[:8], sep="\n")
```
